# Dataset preparation for the Taiwan dataset

This notebook takes initially cleaned Taiwan dataset and prepares train and test datasets

**Author:** Ben Flint  
**Date:** 01/09/2026  
**Input:** taiwan_default_data_vx.x.csv from 00_EDA_Taiwan_dataset.ipynb
**Output:** train and test datasets ready for 03_champion_logistic.ipynb

```python
import pandas as pd
from config import DATA, ID_VARIABLE, RANDOM_SEED, TARGET_VARIABLE
from sklearn.model_selection import train_test_split
```

```python
rds = pd.read_csv(DATA / "processed" / "taiwan_default_data_v0.1.csv")
y = rds[TARGET_VARIABLE]
X = rds.drop([TARGET_VARIABLE, ID_VARIABLE], axis="columns")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.8, random_state=RANDOM_SEED, stratify=y
)
```

```python
output_dir = DATA / "processed" / "basic_data_prep"
output_dir.mkdir(parents=True, exist_ok=True)
for dataset, name in zip(
    [X_train, X_test, y_train, y_test],
    ["X_train", "X_test", "y_train", "y_test"],
    strict=True,
):
    dataset.to_csv(output_dir / f"{name}.csv", index=False)
```
