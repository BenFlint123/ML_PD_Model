# Dataset preparation for the Taiwan dataset

This notebook takes initially cleaned Taiwan dataset and prepares train and test datasets

**Author:** Ben Flint  
**Date:** 01/09/2026  
**Input:** taiwan_default_data_vx.x.csv from 00_EDA_Taiwan_dataset.ipynb
**Output:** train and test datasets ready for 03_champion_logistic.ipynb

```python

import pandas as pd
from config import ID_VARIABLE, TARGET_VARIABLE
from sklearn.model_selection import train_test_split
```

```python
rds = pd.read_csv('../data/processed/taiwan_default_data_v0.1.csv')
y = rds[TARGET_VARIABLE]
X = rds.drop([TARGET_VARIABLE, ID_VARIABLE], axis='columns')
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state = 42, stratify=y)
```

```python
for dataset, name in zip([X_train, X_test, y_train, y_test], ['X_train', 'X_test', 'y_train', 'y_test'], strict=True):
    dataset.to_csv(f"../data/processed/basic_data_prep/{name}.csv", index=False)
```
